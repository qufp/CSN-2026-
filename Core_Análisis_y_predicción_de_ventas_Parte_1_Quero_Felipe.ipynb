```python
import zipfile
import numpy as np

with zipfile.ZipFile("archive (1).zip", "r") as archivo:
    archivo.extract("retail_sales_dataset.csv")

datos = np.genfromtxt(
    "retail_sales_dataset.csv",
    delimiter=",",
    names=True,
    dtype=None,
    encoding="utf-8"
)

print(datos[:5])

print("Cantidad de registros:", len(datos))
```

```python
categorias = np.unique(datos["Product_Category"])

for categoria in categorias:
    filtro = datos["Product_Category"] == categoria
    total_ventas = np.sum(datos["Total_Amount"][filtro])

    print(categoria, ":", total_ventas)
```

```python
fechas = np.unique(datos["Date"])

for categoria in categorias:
    ventas_diarias = []

    for fecha in fechas:
        filtro = (
            (datos["Product_Category"] == categoria) &
            (datos["Date"] == fecha)
        )

        venta_dia = np.sum(datos["Total_Amount"][filtro])
        ventas_diarias.append(venta_dia)

    promedio = np.mean(ventas_diarias)

    print(categoria, ":", round(promedio, 2))
```

```python
totales = np.array([
    np.sum(datos["Total_Amount"][
        datos["Product_Category"] == categoria
    ])
    for categoria in categorias
])

indice_maximo = np.argmax(totales)
indice_minimo = np.argmin(totales)

print("Mayor venta:", categorias[indice_maximo],
      "-", totales[indice_maximo])

print("Menor venta:", categorias[indice_minimo],
      "-", totales[indice_minimo])
```

```python
categoria_seleccionada = "Electronics"

filtro = datos["Product_Category"] == categoria_seleccionada
ventas_filtradas = datos[filtro]

print(ventas_filtradas)

print("Total de ventas:",
      np.sum(ventas_filtradas["Total_Amount"]))

print("Cantidad de transacciones:",
      len(ventas_filtradas))
```

```python
ventas = datos["Total_Amount"]

venta1 = ventas[0]
venta2 = ventas[1]

suma = np.add(venta1, venta2)
resta = np.subtract(venta1, venta2)
multiplicacion = np.multiply(venta1, venta2)
division = np.divide(venta1, venta2)

print("Venta 1:", venta1)
print("Venta 2:", venta2)
print("Suma:", suma)
print("Resta:", resta)
print("Multiplicación:", multiplicacion)
print("División:", division)
print("Suma total:", np.sum(ventas))
print("Promedio por transacción:", np.mean(ventas))
print("Venta máxima:", np.max(ventas))
print("Venta mínima:", np.min(ventas))
print("Mediana:", np.median(ventas))

importe_calculado = (
    datos["Quantity"] * datos["Price_per_Unit"]
)

coinciden = importe_calculado == datos["Total_Amount"]

print("Importes correctos:",
      np.sum(coinciden), "de", len(datos))

print("Importes que no coinciden:",
      np.sum(~coinciden))
```
